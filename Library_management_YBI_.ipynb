## Day 2: Creating Book Records

In this stage, we create a basic structure to store book details.

Each book will contain:
- Book ID
- Book Name
- Author
- Availability Status

```python
books = [
    {
        "id": 1,
        "name": "Python Programming",
        "author": "John Smith",
        "available": True
    },
    {
        "id": 2,
        "name": "Data Structures",
        "author": "Robert Brown",
        "available": True
    }
]

print("Books added successfully!")
```

```python
print("Book List")

for book in books:
    print("ID:", book["id"])
    print("Name:", book["name"])
    print("Author:", book["author"])
    print("Available:", book["available"])
    print("--------------------")
```

## Day 3: Adding New Books

In this stage, the user can add a new book to the library by entering the book ID, book name, and author name.

```python
book_id = int(input("Enter Book ID: "))
book_name = input("Enter Book Name: ")
author_name = input("Enter Author Name: ")

new_book = {
    "id": book_id,
    "name": book_name,
    "author": author_name,
    "available": True
}

books.append(new_book)

print("Book added successfully!")
```

```python
print("Updated Book List")

for book in books:
    print("ID:", book["id"])
    print("Name:", book["name"])
    print("Author:", book["author"])
    print("Available:", book["available"])
    print("--------------------")
```

## Day 4: View Books Functionality

In this stage, a function is created to display all the books available in the library.

```python
def view_books():
    print("\n===== BOOK LIST =====")

    for book in books:
        print("Book ID:", book["id"])
        print("Book Name:", book["name"])
        print("Author:", book["author"])
        print("Status:", "Available" if book["available"] else "Issued")
        print("--------------------")
```

```python
view_books()
```

## Day 5: Search Book Functionality

In this stage, a search function is created to find a book using its Book ID.

```python
def search_book():
    book_id = int(input("Enter Book ID to search: "))

    for book in books:
        if book["id"] == book_id:
            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Book Name:", book["name"])
            print("Author:", book["author"])
            print("Status:", "Available" if book["available"] else "Issued")
            return

    print("Book not found.")
```

```python
search_book()
```

## Day 6: Issue Book Functionality

In this stage, a function is created to issue a book from the library. The availability status of the book is updated after issuing.

```python
def issue_book():
    book_id = int(input("Enter Book ID to issue: "))

    for book in books:
        if book["id"] == book_id:
            if book["available"]:
                book["available"] = False
                print("Book issued successfully!")
            else:
                print("Book is already issued.")
            return

    print("Book not found.")
```

```python
issue_book()
```

## Day 7: Return Book Functionality

In this stage, a function is created to return an issued book. The availability status of the book is updated after returning it.

```python
def return_book():
    book_id = int(input("Enter Book ID to return: "))

    for book in books:
        if book["id"] == book_id:
            if not book["available"]:
                book["available"] = True
                print("Book returned successfully!")
            else:
                print("Book is already available.")
            return

    print("Book not found.")
```

```python
return_book()
```

```python
view_books()
```

## Day 8: Add Book Function

In this stage, the book adding process is converted into a function.
This makes the program more organized and easier to use.

```python
def add_book():
    book_id = int(input("Enter Book ID: "))
    book_name = input("Enter Book Name: ")
    author_name = input("Enter Author Name: ")

    new_book = {
        "id": book_id,
        "name": book_name,
        "author": author_name,
        "available": True
    }

    books.append(new_book)

    print("Book added successfully!")
```

```python
add_book()
```

```python
view_books()
```

## Day 9: Creating the Main Menu

In this stage, a menu is created to access different library management operations. The user can select an option to add, view, search, issue, or return books.


```python
def library_menu():
    print("\n===== LIBRARY MANAGEMENT SYSTEM =====")
    print("1. Add Book")
    print("2. View Books")
    print("3. Search Book")
    print("4. Issue Book")
    print("5. Return Book")
    print("6. Exit")
```

```python
library_menu()
```

## Day 10: Connecting Menu with Functions

In this stage, the main menu is connected with library functions. The user can select an option to perform different library operations.

```python
def run_library_menu():
    library_menu()
    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()
    elif choice == "2":
        view_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        issue_book()
    elif choice == "5":
        return_book()
    elif choice == "6":
        print("Thank you for using the Library Management System!")
    else:
        print("Invalid choice! Please enter a number from 1 to 6.")
```

```python
run_library_menu()
```

In this stage, input validation is added to the library menu. If the user enters an invalid option, the program displays an appropriate message.

```python
run_library_menu()
```

## Day 12: Book Availability Report

In this section, we will display the availability status of all books in the library.

```python
print("Library Book Availability Report")
print("-" * 35)

for book in books:
    if book["available"]:
        status = "Available"
    else:
        status = "Issued"

    print("Book ID:", book["id"])
    print("Book Name:", book["name"])
    print("Author:", book["author"])
    print("Status:", status)
    print("-" * 35)
```

## Day 13: Introduction to File Handling

## Day 14: Saving Book Records to a File

In this stage, book records are saved to a text file using Python file handling. This helps store book information outside the program.

```python
def save_books():
    with open("library.txt", "w") as file:
        for book in books:
            file.write(
                f'{book["id"]},{book["name"]},'
                f'{book["author"]},{book["available"]}\n'
            )

    print("Book records saved successfully!")

save_books()
```

```python
with open("library.txt", "r") as file:
    print(file.read())
```

## Day 15: Search Book by Name

In this section, we will implement a book search feature that allows users to find books by their names in the library.

```python
def search_book():
    search_name = input("Enter book name to search: ").lower()
    found = False

    for book in books:
        if search_name in book["name"].lower():
            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Book Name:", book["name"])
            print("Author:", book["author"])
            print("Status:", "Available" if book["available"] else "Issued")
            found = True

    if not found:
        print("Book not found in the library.")

search_book()
```

## Day 16: Count Available and Issued Books

In this section, we will count the total number of available and issued books in the library using Python.

```python
available_books = 0
issued_books = 0

for book in books:
    if book["available"]:
        available_books += 1
    else:
        issued_books += 1

print("Library Book Summary")
print("--------------------")
print("Total Books:", len(books))
print("Available Books:", available_books)
print("Issued Books:", issued_books)
```

## Day 17: Sort Books by Name

In this section, we will sort the books in alphabetical order by their names using Python.

```python
sorted_books = sorted(books, key=lambda book: book["name"].lower())

print("Library Books in Alphabetical Order")
print("-" * 40)

for book in sorted_books:
    print("Book ID:", book["id"])
    print("Book Name:", book["name"])
    print("Author:", book["author"])
    print("Status:", "Available" if book["available"] else "Issued")
    print("-" * 40)
```

## Day 18: Search Books by Author

In this section, we will implement a feature to search for books by author name in the Library Management System.

```python
def search_by_author():
    author_name = input("Enter author name: ").strip().lower()
    found = False

    for book in books:
        if author_name in book["author"].lower():
            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Book Name:", book["name"])
            print("Author:", book["author"])
            print("Status:", "Available" if book["available"] else "Issued")
            print("-" * 30)
            found = True

    if not found:
        print("No books found for this author.")

search_by_author()
```
