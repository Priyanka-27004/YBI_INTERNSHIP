## Day 3: Adding New Books

In this stage, the user can add a new book to the library by entering the book ID, book name, and author name.

```python
book_id = int(input("Enter Book ID: "))
book_name = input("Enter Book Name: ")
author_name = input("Enter Author Name: ")

new_book = {
    "id": book_id,
    "name": book_name,
    "author": author_name,
    "available": True
}

books.append(new_book)

print("Book added successfully!")
```

## Day 4: View Books Functionality

In this stage, a function is created to display all the books available in the library.

```python
def view_books():
    print("\n===== BOOK LIST =====")

    for book in books:
        print("Book ID:", book["id"])
        print("Book Name:", book["name"])
        print("Author:", book["author"])
        print("Status:", "Available" if book["available"] else "Issued")
        print("--------------------")
```

```python
view_books()
```

## Day 5: Search Book Functionality

In this stage, a search function is created to find a book using its Book ID.

```python
def search_book():
    book_id = int(input("Enter Book ID to search: "))

    for book in books:
        if book["id"] == book_id:
            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Book Name:", book["name"])
            print("Author:", book["author"])
            print("Status:", "Available" if book["available"] else "Issued")
            return

    print("Book not found.")
```

```python
search_book()
```

## Day 6: Issue Book Functionality

In this stage, a function is created to issue a book from the library. The availability status of the book is updated after issuing.

```python
def issue_book():
    book_id = int(input("Enter Book ID to issue: "))

    for book in books:
        if book["id"] == book_id:
            if book["available"]:
                book["available"] = False
                print("Book issued successfully!")
            else:
                print("Book is already issued.")
            return

    print("Book not found.")
```

```python
issue_book()
```

## Day 7: Return Book Functionality

In this stage, a function is created to return an issued book. The availability status of the book is updated after returning it.

```python
def return_book():
    book_id = int(input("Enter Book ID to return: "))

    for book in books:
        if book["id"] == book_id:
            if not book["available"]:
                book["available"] = True
                print("Book returned successfully!")
            else:
                print("Book is already available.")
            return

    print("Book not found.")
```

```python
return_book()
```

## Day 14: Saving Book Records to a File

In this stage, book records are saved to a text file using Python file handling. This helps store book information outside the program.

```python
def save_books():
    with open("library.txt", "w") as file:
        for book in books:
            file.write(
                f'{book["id"]},{book["name"]},'
                f'{book["author"]},{book["available"]}\n'
            )

    print("Book records saved successfully!")

save_books()
```

## Day 15: Search Book by Name

In this section, we will implement a book search feature that allows users to find books by their names in the library.

```python
def search_book():
    search_name = input("Enter book name to search: ").lower()
    found = False

    for book in books:
        if search_name in book["name"].lower():
            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Book Name:", book["name"])
            print("Author:", book["author"])
            print("Status:", "Available" if book["available"] else "Issued")
            found = True

    if not found:
        print("Book not found in the library.")

search_book()
```

## Day 18: Search Books by Author

In this section, we will implement a feature to search for books by author name in the Library Management System.

```python
def search_by_author():
    author_name = input("Enter author name: ").strip().lower()
    found = False

    for book in books:
        if author_name in book["author"].lower():
            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Book Name:", book["name"])
            print("Author:", book["author"])
            print("Status:", "Available" if book["available"] else "Issued")
            print("-" * 30)
            found = True

    if not found:
        print("No books found for this author.")

search_by_author()
```

## Day 19: Update Book Availability

In this section, we will implement a feature to update the availability status of a book when it is issued or returned.

```python
def update_book_status():
    book_id = input("Enter Book ID: ").strip()
    found = False

    for book in books:
        if str(book["id"]) == book_id:
            found = True

            print("Book Name:", book["name"])
            print("Current Status:",
                  "Available" if book["available"] else "Issued")

            choice = input("Enter 1 to Issue or 2 to Return: ").strip()

            if choice == "1":
                if book["available"]:
                    book["available"] = False
                    print("Book issued successfully!")
                else:
                    print("This book is already issued.")

            elif choice == "2":
                if not book["available"]:
                    book["available"] = True
                    print("Book returned successfully!")
                else:
                    print("This book is already available.")

            else:
                print("Invalid choice.")

            break

    if not found:
        print("Book ID not found.")


update_book_status()
```

## Day 20: Library Fine Calculator

In this section, we will calculate the library fine for books returned after the due date. A fine of ₹5 per day will be charged for late returns.

```python
def calculate_fine():
    late_days = input("Enter number of late days: ").strip()

    if not late_days.isdigit():
        print("Please enter a valid number of days.")
        return

    late_days = int(late_days)
    fine_per_day = 5
    total_fine = late_days * fine_per_day

    print("\nLibrary Fine Details")
    print("--------------------")
    print("Late Days:", late_days)
    print("Fine Per Day: Rs.", fine_per_day)
    print("Total Fine: Rs.", total_fine)


calculate_fine()
```

## Day 21: Library Statistics Report

In this section, we will generate a library statistics report showing the total number of books, available books, issued books, and the percentage of available books.

```python
def library_statistics():
    total_books = len(books)
    available_books = sum(1 for book in books if book["available"])
    issued_books = total_books - available_books

    if total_books > 0:
        availability_percentage = (available_books / total_books) * 100
    else:
        availability_percentage = 0

    print("===== Library Statistics Report =====")
    print("Total Books:", total_books)
    print("Available Books:", available_books)
    print("Issued Books:", issued_books)
    print("Availability Percentage:", round(availability_percentage, 2), "%")
    print("=====================================")

library_statistics()
```

## Day 22: Export Library Records to CSV

In this section, we will export library book records into a CSV file using Python's built-in CSV module. This allows library data to be stored and viewed in spreadsheet applications.

```python
import csv

def export_books_to_csv():
    if not books:
        print("No books available to export.")
        return

    with open("library_books.csv", "w", newline="", encoding="utf-8") as file:
        fieldnames = ["id", "name", "author", "available"]
        writer = csv.DictWriter(file, fieldnames=fieldnames)

        writer.writeheader()
        writer.writerows(books)

    print("Library records exported successfully!")
    print("File name: library_books.csv")

export_books_to_csv()
```

## Day 23: Read Library Records from CSV

In this section, we will read and display book records stored in a CSV file using Python's built-in CSV module.

```python
import csv
import os

def read_books_from_csv():
    filename = "library_books.csv"

    if not os.path.exists(filename):
        print("CSV file not found. Please export books first.")
        return

    with open(filename, "r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        print("===== Library Records from CSV =====")

        found = False

        for book in reader:
            found = True
            print("Book ID:", book["id"])
            print("Book Name:", book["name"])
            print("Author:", book["author"])
            print(
                "Status:",
                "Available" if book["available"].lower() == "true"
                else "Issued"
            )
            print("-" * 30)

        if not found:
            print("No book records found.")

read_books_from_csv()
```

## Day 24: Backup Library Records Using JSON

In this section, we will save library book records into a JSON file using Python. This helps us keep a backup of the library data.

```python
import json

def backup_library():
    if not books:
        print("No books available to back up.")
        return

    with open("library_backup.json", "w", encoding="utf-8") as file:
        json.dump(books, file, indent=4)

    print("Library backup created successfully!")
    print("File name: library_backup.json")

backup_library()
```

# Day 25: Restore Library Records from JSON

**Objective:** To restore library book records from a JSON backup file using Python.

**Description:** This program loads previously saved book records from a JSON file and displays them.

```python

import json
import os

backup_file = "library_backup.json"

if os.path.exists(backup_file):
    with open(backup_file, "r") as file:
        books = json.load(file)

    print("Library records restored successfully!")
    print("\nLibrary Book Records:")

    for book in books:
        print(book)
else:
    print("Backup file not found. Please create a backup first.")
```

# Day 26: Search Book by Title

**Objective:** To search for a book by its title using Python.

**Description:** This program allows users to search for a specific book in the library records and displays its details if found.

```python

books = [
    {"id": 1, "title": "Python Programming", "status": "Available"},
    {"id": 2, "title": "Data Structures", "status": "Issued"},
    {"id": 3, "title": "Web Development", "status": "Available"}
]

search_title = input("Enter book title to search: ").strip().lower()

found = False

for book in books:
    if search_title == book["title"].lower():
        print("\nBook Found!")
        print("Book ID:", book["id"])
        print("Title:", book["title"])
        print("Status:", book["status"])
        found = True
        break

if not found:
    print("Book not found in the library.")
```
